# ImageToolkit Notebook
This notebook contains demonstrations for image processing operations.

```python
import cv2, numpy as np
import matplotlib.pyplot as plt
from PIL import Image
plt.rcParams['figure.figsize'] = (8,6)
```

```python
# Create synthetic example
h,w=256,256
x = np.linspace(0,255,w,np.uint8)
y = np.linspace(0,255,h,np.uint8)[:,None]
img = np.dstack([(x+y)//2, np.flip(x)[None].repeat(h,0), np.full((h,w),128,np.uint8)])
plt.imshow(img); plt.axis('off'); plt.title('Synthetic Example'); plt.show()
```

## Color conversion example
Convert to grayscale and show.

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(gray, cmap='gray'); plt.axis('off'); plt.title('Gray'); plt.show()
```
